# OASIS Simulation - JKN Fraud Patterns

This notebook runs OASIS simulations to analyze JKN fraud patterns.

```python
import sys
sys.path.append('../backend/src')

import asyncio
import json
from oasis.simulation import JKNFraudSimulation
```

```python
# Initialize simulation
simulation = JKNFraudSimulation(
    openai_api_base="https://9router.biezz.my.id/v1",
    openai_api_key="sk-dc3493f22a937c2c-qwa9dk-fe62b641",
    model_name="oasis"
)

print("Simulation initialized!")
```

```python
# Run phantom billing simulation
async def run_phantom():
    results = await simulation.run_phantom_billing_simulation(num_agents=50, num_steps=10)
    print(json.dumps(results, indent=2))
    return results

phantom_results = await run_phantom()
```

```python
# Run upcoding simulation
async def run_upcoding():
    results = await simulation.run_upcoding_simulation(num_agents=50, num_steps=10)
    print(json.dumps(results, indent=2))
    return results

upcoding_results = await run_upcoding()
```

```python
# Run collusion simulation
async def run_collusion():
    results = await simulation.run_collusion_simulation(num_agents=50, num_steps=10)
    print(json.dumps(results, indent=2))
    return results

collusion_results = await run_collusion()
```

```python
# Run all simulations
async def run_all():
    results = await simulation.run_all_simulations(num_agents=100, num_steps=10)
    print(json.dumps(results, indent=2))
    return results

all_results = await run_all()
```

```python
# Compare with baseline
comparison = simulation.compare_with_baseline(all_results, baseline_auprc=0.71)
print(json.dumps(comparison, indent=2))
```

```python
# Visualize results
import matplotlib.pyplot as plt

simulations = all_results['simulations']
names = ['Phantom Billing', 'Upcoding', 'Collusion']
detection_rates = [
    simulations['phantom_billing']['detection_rate'],
    simulations['upcoding']['detection_rate'],
    simulations['collusion']['detection_rate']
]

plt.figure(figsize=(10, 6))
plt.bar(names, detection_rates, color=['#1f77b4', '#ff7f0e', '#2ca02c'])
plt.axhline(y=0.71, color='r', linestyle='--', label='Baseline (XGBoost)')
plt.xlabel('Fraud Pattern')
plt.ylabel('Detection Rate')
plt.title('OASIS Detection Rates by Fraud Pattern')
plt.legend()
plt.ylim(0, 1)
plt.show()
```
